```python
# ============================================
# PHASE 2 — Cross-Dataset Consistency (RQ2)
# Spearman's Rank Correlation across NSL-KDD, UNSW-NB15, CICIDS2017
# ============================================

import pandas as pd
from scipy.stats import spearmanr
from itertools import combinations

# ---- 1. LOAD PHASE 1 SUMMARY RESULTS FOR ALL THREE DATASETS ----
nslkdd_summary = pd.read_csv(
    r"C:\Users\icevo\OneDrive\Desktop\Dissertation\Datasets\NSL-KDD\results\Phase1_NSLKDD_summary.csv",
    index_col=0
)
unsw_summary = pd.read_csv(
    r"C:\Users\icevo\OneDrive\Desktop\Dissertation\Datasets\UNSW-NB15\results\Phase1_UNSWNB15_summary.csv",
    index_col=0
)
cicids_summary = pd.read_csv(
    r"C:\Users\icevo\OneDrive\Desktop\Dissertation\Datasets\CIC-IDS2017\results\Phase1_CICIDS2017_summary.csv",
    index_col=0
)

print("NSL-KDD summary:\n", nslkdd_summary, "\n")
print("UNSW-NB15 summary:\n", unsw_summary, "\n")
print("CICIDS2017 summary:\n", cicids_summary, "\n")

# ---- 2. BUILD RANKINGS BASED ON MCC (my chosen primary metric) ----
# Rank 1 = best (highest MCC)
nslkdd_rank = nslkdd_summary['mcc'].rank(ascending=False)
unsw_rank = unsw_summary['mcc'].rank(ascending=False)
cicids_rank = cicids_summary['mcc'].rank(ascending=False)

# Combine into one table for clarity — algorithms as rows, datasets as columns
rankings = pd.DataFrame({
    'NSL-KDD': nslkdd_rank,
    'UNSW-NB15': unsw_rank,
    'CICIDS2017': cicids_rank
})

print("Algorithm rankings by dataset (1 = best):\n")
print(rankings)

# ---- 3. SPEARMAN'S RANK CORRELATION — PAIRWISE BETWEEN DATASETS ----
datasets = {
    'NSL-KDD': nslkdd_rank,
    'UNSW-NB15': unsw_rank,
    'CICIDS2017': cicids_rank
}

print("\nSpearman's Rank Correlation (pairwise between datasets):\n")

correlation_results = []

for dataset_a, dataset_b in combinations(datasets.keys(), 2):
    rho, p_value = spearmanr(datasets[dataset_a], datasets[dataset_b])
    significant = "SIGNIFICANT (p < 0.05)" if p_value < 0.05 else "not significant"
    print(f"{dataset_a} vs {dataset_b}: rho={rho:.4f}, p={p_value:.4f} -> {significant}")
    correlation_results.append({
        'Dataset A': dataset_a,
        'Dataset B': dataset_b,
        'Spearman rho': round(rho, 4),
        'p-value': round(p_value, 4),
        'Significant': significant
    })

correlation_df = pd.DataFrame(correlation_results)

# ---- 4. SAVE RESULTS ----
import os
results_folder = r"C:\Users\icevo\OneDrive\Desktop\Dissertation\Datasets\Phase2_results"
os.makedirs(results_folder, exist_ok=True)

rankings.to_csv(os.path.join(results_folder, "Phase2_algorithm_rankings.csv"))
correlation_df.to_csv(os.path.join(results_folder, "Phase2_spearman_correlations.csv"), index=False)

print("\nResults saved successfully!")
print("Rankings:", os.path.join(results_folder, "Phase2_algorithm_rankings.csv"))
print("Correlations:", os.path.join(results_folder, "Phase2_spearman_correlations.csv"))
```
